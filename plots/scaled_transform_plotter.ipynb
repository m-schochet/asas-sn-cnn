```python
from astropy.table import Table
from common_functions import single_wavelet, scaled_wavelet, inject_flux, read_sim, plotter, reduce
import glob
import numpy as np
import matplotlib.colors as colors
import os
import pathlib
from pyasassn.client import SkyPatrolClient
import matplotlib.pyplot as plt
```

```python
os.chdir(pathlib.Path.cwd())
def set_rcparams():
    tab = Table.read('rcparams.txt', format='csv')
    for i in range(len(tab)):
        try:
            plt.rcParams[tab['key'][i]] = float(tab['val'][i])
        except ValueError:
            plt.rcParams[tab['key'][i]] = str(tab['val'][i])
    return
set_rcparams()
```

```python
client = SkyPatrolClient()
```

```python
def scaledpipeline(objid, flux=["noninje"]):
    lcs = client.query_list([objid], download=True)
    for lc in lcs.itercurves():
        savedlc = lc
    savedlc_fin = reduce(savedlc)
    if(len(flux)!=1):
        wavelet = scaled_wavelet(savedlc_fin, flux)
        plotter(wavelet, True, 'yes')
    else:
        wavelet = scaled_wavelet(savedlc_fin, savedlc_fin.data.flux)
        plotter(wavelet)
    return wavelet

def regpipeline(objid, flux=["noninje"]):
    lcs = client.query_list([objid], download=True)
    for lc in lcs.itercurves():
        savedlc = lc
    savedlc_fin = reduce(savedlc)
    if(len(flux)!=1):
        wavelet = single_wavelet(savedlc_fin, flux)
        plotter(wavelet, True)
    else:
        wavelet = single_wavelet(savedlc_fin, savedlc_fin.data.flux)    
        plotter(wavelet)
    return savedlc_fin
```

```python
savedlcred = regpipeline(661427253681)
```

```python
simulation_file = glob.glob("files/sim900184.fits")
simulation = read_sim(simulation_file[0])
newflux = inject_flux(simulation, savedlcred)
```

```python
scaled = scaledpipeline(661427253681, newflux)
```

```python
notscaled = regpipeline(661427253681, newflux)
```
