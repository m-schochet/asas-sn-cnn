```python
from astropy.table import Table
import requests
```

```python
urlsantos1 = "https://content.cld.iop.org/journals/0067-0049/244/1/21/revision1/apjsab3b56t3_mrt.txt"
urlsantos2 = "https://content.cld.iop.org/journals/0067-0049/255/1/17/revision1/apjsac033ft1_mrt.txt"
urlluztf = "https://content.cld.iop.org/journals/1538-3881/164/6/251/revision1/ajac9beet1_mrt.txt"
urlclaytortess = "https://content.cld.iop.org/journals/0004-637X/962/1/47/revision1/apjad159at5_mrt.txt"
urlclaytorkbonus = "https://content.cld.iop.org/journals/0004-637X/987/1/8/revision1/apjadd5f0t1_mrt.txt"
urlleinerssgs = "https://content.cld.iop.org/journals/0004-637X/927/2/222/revision1/apjac53b1t1_mrt.txt"
query_parameters = {"downloadformat": "txt"}
```

```python
response = requests.get(urlsantos1, params=query_parameters)
with open("santos1.txt", mode="wb") as file:
    file.write(response.content)
santos1 = Table.read("santos1.txt", format="ascii.cds").to_pandas()
```
